```python
# Import everything needed and create the stemmer
import re
import spacy
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
nlp = spacy.load('en_core_web_sm')
```

### Q1
Stem these related words and print the result for each one.

```python
# Stem a group of related words
words = ['connect', 'connected', 'connecting', 'connection', 'connections']

for word in words:
  print(word, "->", stemmer.stem(word))
```

### Q2
Take a short sentence, split it into words, and stem every word.

```python
# Split a sentence and stem each word
text = 'The students are studying different subjects'
words = text.split()

for word in words:
  print(word, "->", stemmer.stem(word))
```

### Q3
Stem these adjectives that show different degrees of happiness.

```python
# Stem words related to happiness
words = ['happy', 'happier', 'happiest', 'happiness']

for word in words:
  stem = stemmer.stem(word)
  print(f"{word:12} -> {stem}")
```

### Q4
Stem these organization-related words.

```python
# Stem organization-related words
words = ['organize', 'organizing', 'organization', 'organizations']

for word in words:
  print(word, "->", stemmer.stem(word))
```

### Q5
Convert the whole sentence to lowercase, split it, stem every word, and then join the stemmed words back into a new sentence.

```python
# Stem a full sentence and rebuild it
text = 'The players are playing football and practising regularly.'
words = text.lower().split()
stemmed_words = []

for word in words:
  stemmed_words.append(stemmer.stem(word))

stemmed_text = ' '.join(stemmed_words)

print('Original:')
print(text)

print("\nStemmed:")
print(stemmed_text)
```

### Q6
Tokenize the sentence using spaCy and stem only the alphabetic tokens.

```python
# Tokenize with spaCy and stem alphabetic tokens
text = 'the students are studying'
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

### Q7
Stem all the alphabetic words that appear in these three short sentences about playing.

```python
# Stem forms of play across multiple sentences
text = 'I play football.He plays football.They played yesterday.'
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

### Q8
Stem the tokens from this short paragraph and print them in aligned columns.

```python
# Stem tokens and print in aligned format
text = '''Data scientists analyse data.
They are analyzing datasets and developed useful models.'''

doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

### Q9
First convert the tokens to lowercase, then stem them.

```python
# Lowercase the tokens first, then stem
text = 'Python is powerful,useful, and popular!'
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    word = token.text.lower()
    print(word, "->", stemmer.stem(word))
```

### Q10
Tokenize the sentence, keep only alphabetic tokens, stem them after converting to lowercase, and rebuild the full stemmed sentence.

```python
# Stem a sentence and rebuild it from the stemmed words
text = 'The children are playing games and enjoying themselves'

doc = nlp(text)

stemmed_words = []
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text = " ".join(stemmed_words)

print('Original:')
print(text)

print("\nStemmed:")
print(stemmed_text)
```
