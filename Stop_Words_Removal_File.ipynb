# Part A: spaCy (Questions 1 to 5)

### Question 1
Find out how many words are present in spaCy's default stop-word list.

```python
#Display spacy's stopword list
import spacy
nlp=spacy.load('en_core_web_sm')
print(len(nlp.Defaults.stop_words))
```

### Question 2
Remove the stop words from the sentence `'The cute cat and the cute dog are playing in the garden'` and print the remaining words as one sentence.

```python
# Remove stop words and join the remaining words
text='The cute cat and the cute dog are playing in the garden'
doc=nlp(text)
result=[]


for token in doc:
  if not token.is_stop:
    result.append(token.text)

print(" ".join(result))
```

### Question 3
Remove both stop words and punctuation from the sentence `'The cute cat is sleeping on the sofa, and the cute dog is barking happily.'` and print the cleaned sentence.

```python
# Remove stop words and punctuation, then join the words
text='The cute cat is sleeping on the sofa, and the cute dog is barking happily.'
doc=nlp(text)
result=[]


for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)

print(" ".join(result))
```

### Question 4
Add the word `'cat'` to spaCy's stop-word list, then check the stop-word status of every token in `'Cat is a cute and friendly pet'`.

```python
# Add a custom stop word and check token status
import spacy
nlp=spacy.load('en_core_web_sm')
nlp.Defaults.stop_words.add('cat')
doc=nlp('Cat is a cute and friendly pet')
for token in doc:
  print(token.text,"->",token.is_stop)
```

### Question 5
Define a function `remove_stopwords(text)` that removes stop words and punctuation, then call it on `"The cute cats are chasing Little Brown Dogs"`.

```python
# Define a function to remove stop words and punctuation
def remove_stopwords(text):
    doc = nlp(text)
    result = []

    for token in doc:
        if not token.is_stop and not token.is_punct:
            result.append(token.text)

    return " ".join(result)


text = "The cute cats are chasing Little Brown Dogs"

print(remove_stopwords(text))
```

### Question 7
Create the NLTK English stop-word set and print how many words it contains.

### Question 8
Check whether each word in the list `['The','Is','Cat','And','cute']` is present in the NLTK stop-word set. The list contains capitalised words, so think about case.

```python
# Check each word against the stop-word set
stop_words=set(stopwords.words('english'))
words=['The','Is','Cat','And','cute']

for word in words:
  print(word,"->",word.lower() in stop_words)
```

### Question 9
Tokenize the sentence `'A cute dog is a very loyal friend'` and print only the stop words found in it.

```python
# Tokenize the text and extract only the stop words
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
stop_words=set(stopwords.words('english'))

text='A cute dog is a very loyal friend'

tokens=word_tokenize(text)
result=[]
for word in tokens:
  if word.lower() in stop_words:
    result.append(word)

print(result)
```

### Question 10
Tokenize the sentence `'The cute cat and the cute dog are eating fish, chicken, and rice.'` and remove both stop words and punctuation tokens.

```python
# Remove stop words and punctuation from the tokens
import nltk
nltk.download('stopwords')


from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

nltk.download('punkt')
nltk.download('punkt_tab')
stop_words=set(stopwords.words('english'))

text='The cute cat and the cute dog are eating fish, chicken, and rice.'

tokens=word_tokenize(text)
result=[]
for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)

print(result)
```
