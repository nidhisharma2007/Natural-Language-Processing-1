## Q1. Practise file operations and display section headings

**(a)** Open `machine_learning.txt`, read its lines, and print the section headings that appear in the file.
**(b)** Copy the contents of `machine_learning.txt` into `text_file2.txt` using `w+`. Print the number of characters written, then use `seek(0)` to read the copied text.
**(c)** Read the first 100 characters of `text_file2.txt`, move the cursor to the beginning with `seek(0)`, and read the first 100 characters again. Close the file.
**(d)** Append a new line to `text_file2.txt` without replacing its existing content.
**(e)** Print `text_file2.txt` line by line and close the file.

```python
# Q1
with open('machine_learning.txt') as f:
    lines = f.readlines()

headings = ['Introduction', 'History of Machine Learning',
            'Types of Machine Learning', 'Common Algorithms',
            'The Machine Learning Workflow',
            'Machine Learning and Natural Language Processing',
            'Evaluating a Model', 'Applications of Machine Learning',
            'Challenges and Ethics', 'Conclusion']

for line in lines:
    if line.strip() in headings:
        print(line.strip())
```

**Q1(b) — Copy the source file into `text_file2.txt` using `w+`.**

```python
with open('machine_learning.txt') as f:
    text = f.read()

my_file = open('text_file2.txt', 'w+')
characters_written = my_file.write(text)
print('Characters written:', characters_written)
my_file.seek(0)
print(my_file.read())
my_file.close()
```

**Q1(c) — Practise reading and moving the file cursor.**

```python
my_file = open('text_file2.txt', 'r')
print(my_file.read(100))
my_file.seek(0)
print(my_file.read(100))
my_file.close()
```

**Q1(d) — Append a line without overwriting the copied text.**

```python
my_file = open('text_file2.txt', 'a+')
my_file.write('\nThis is an extra practice line.')
my_file.close()
```

**Q1(e) — Read the updated file line by line.**

```python
my_file = open('text_file2.txt', 'r')
for line in my_file:
    print(line, end="")
my_file.close()
```

## Q4. Create a vocabulary

**(a)** Make a list of unique alphabetic words from the file. Store the words in lowercase.

**(b)** Print the first 40 vocabulary words and the total vocabulary size.


```python
# Q4
with open('machine_learning.txt') as f:
    text = f.read()

doc = nlp(text)
vocabulary = []

for token in doc:
    if token.is_alpha and token.text.lower() not in vocabulary:
        vocabulary.append(token.text.lower())

print(vocabulary[:40])
print('Vocabulary size:', len(vocabulary))
```

## Q5. Match machine-learning algorithm names

**(a)** Use spaCy `Matcher` to find the algorithm phrases listed in the patterns.

**(b)** Print each matched phrase from `machine_learning.txt`.


```python
# Q5
from spacy.matcher import Matcher

with open('machine_learning.txt') as f:
    text = f.read()

doc = nlp(text)
matcher = Matcher(nlp.vocab)

pattern1 = [{'LOWER': 'linear'}, {'LOWER': 'regression'}]
pattern2 = [{'LOWER': 'logistic'}, {'LOWER': 'regression'}]
pattern3 = [{'LOWER': 'random'}, {'LOWER': 'forest'}]
pattern4 = [{'LOWER': 'neural'}, {'LOWER': 'networks'}]

matcher.add('ALGORITHMS', [pattern1, pattern2, pattern3, pattern4])

matches = matcher(doc)

for match_id, start, end in matches:
    print(doc[start:end].text)
```

## Q6. Stop words in the paragraph about privacy

**(a)** Find the paragraph containing `Privacy is another concern` and print its stop words.

**(b)** From that same paragraph, print the words after removing stop words and punctuation.


```python
# Q6(a)
with open('machine_learning.txt') as f:
    lines = f.readlines()

for line in lines:
    if 'Privacy is another concern' in line:
        doc = nlp(line)
        for token in doc:
            if token.is_stop:
                print(token.text)
```

```python
# Q6(b)
with open('machine_learning.txt') as f:
    lines = f.readlines()

for line in lines:
    if 'Privacy is another concern' in line:
        doc = nlp(line)
        for token in doc:
            if not token.is_stop and not token.is_punct and not token.is_space:
                print(token.text)
```

## Q7. Lemmatize verbs in the file

**(a)** Read `machine_learning.txt` and create a spaCy `Doc`.

**(b)** Print each verb together with its lemma.


```python
# Q7
with open('machine_learning.txt') as f:
    text = f.read()

doc = nlp(text)

for token in doc:
    if token.pos_ == 'VERB':
        print(token.text, token.lemma_)
```

## Q8. Find nouns whose form changes after lemmatization

**(a)** Read the file and process it with spaCy.

**(b)** Print nouns whose original form differs from their lemma.


```python
# Q8
with open('machine_learning.txt') as f:
    text = f.read()

doc = nlp(text)

for token in doc:
    if token.pos_ == 'NOUN' and token.text.lower() != token.lemma_.lower():
        print(token.text, token.lemma_)
```

## Q11. Recognise dates and money

**(a)** Read the text file and create a spaCy `Doc`.

**(b)** Print named entities labelled `DATE`.

**(c)** Print named entities labelled `MONEY`.


```python
# Q11
with open('machine_learning.txt') as f:
    text = f.read()

doc = nlp(text)

print('Dates:')
for ent in doc.ents:
    if ent.label_ == 'DATE':
        print(ent.text, ent.label_)

print('Money:')
for ent in doc.ents:
    if ent.label_ == 'MONEY':
        print(ent.text, ent.label_)
```
