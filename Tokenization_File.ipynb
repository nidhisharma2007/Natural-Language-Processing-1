# Part A – NLTK Questions

## Question 1
Tokenize the sentence and also print how many tokens are there.

```python
# Tokenize the sentence and count the tokens
from nltk.tokenize import word_tokenize

text = "cats are cute and dogs are loyal"
tokens = word_tokenize(text)

print('Tokens:', tokens)
print('Number of tokens:', len(tokens))
```

## Question 2
Tokenize a sentence that has a contraction like "can't" and see how NLTK splits it.

```python
# Observe how NLTK handles contractions
from nltk.tokenize import word_tokenize

text = "the dog can't catch the cat"
tokens = word_tokenize(text)

print(tokens)
```

## Question 3
Use `wordpunct_tokenize` on the given text and print the tokens.

```python
# Tokenize using wordpunct_tokenize
from nltk.tokenize import wordpunct_tokenize

text = "Wow! The cat is fast. Isn't the dog faster?"
tokens = wordpunct_tokenize(text)

print(tokens)
```

## Question 4
Tokenize the text, remove all punctuation marks, and print both the original tokens and the cleaned words.

```python
# Remove punctuation after tokenization
import string
from nltk.tokenize import word_tokenize

text = "cats are playful and dogs are friendly."
tokens = word_tokenize(text)
words = [token for token in tokens if token not in string.punctuation]

print('All tokens:', tokens)
print('Words:', words)
print('Number of words:', len(words))
```

## Question 5
Convert all the tokens to lowercase and print both the original and the lowercase versions.

```python
# Convert tokens to lowercase
from nltk.tokenize import word_tokenize

text = "CAT CAT CAT is sleeping."
tokens = word_tokenize(text)
lower_tokens = [word.lower() for word in tokens]

print('Original tokens:')
print(tokens)
print('\nLowercase tokens:')
print(lower_tokens)
```

# Part B – spaCy Questions

## Question 6
Tokenize a sentence that has numbers and a currency symbol, then print each token.

```python
# Tokenize text containing numbers and currency
text = 'I bought 3 cats and 2 dogs for $150'
doc = nlp(text)

for token in doc:
  print(token.text)
```

## Question 7
Tokenize a sentence that has a contraction and see how spaCy splits it.

```python
# Observe how spaCy handles contractions
text = "I don't like mean dogs."
doc = nlp(text)

for token in doc:
  print(token.text)
```

## Question 8
Tokenize a sentence that has hyphenated words and print each token.

```python
# Tokenize hyphenated words
text = "Cat-loving and dog-friendly people are kind"
doc = nlp(text)

for token in doc:
  print(token.text)
```

## Question 9
After tokenization, print only the alphabetic tokens (use `token.is_alpha`).

```python
# Keep only alphabetic tokens
text = "My cat is 2 years old!"
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(token.text)
```

## Question 10
Split the text into sentences. For each sentence, print the sentence and then print all its tokens.

```python
# Sentence segmentation followed by word tokenization
text = 'The cat is sleeping on the sofa. The dog is playing in the garden.'
doc = nlp(text)

for sentence in doc.sents:
  print('Sentence:', sentence.text)
  for token in sentence:
    print(token.text)
  print()
```
