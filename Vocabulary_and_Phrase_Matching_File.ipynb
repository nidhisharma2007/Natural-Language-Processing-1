```python
# Import spaCy and load the model
import spacy
nlp=spacy.load('en_core_web_sm')
```

```python
# Create a Matcher object
from spacy.matcher import Matcher
matcher=Matcher(nlp.vocab)
```

```python
# Create a doc to search in
doc=nlp(u'Puppy Love is cute. I love puppylove and Puppy-love')
```

---
# Questions

### Question 1
Add patterns to the matcher to find `puppylove`, `puppy love` and `puppy-love`. Name the rule `PuppyLove`.

```python
# Add three patterns for puppy love
pattern1=[{'LOWER':'puppylove'}]
pattern2=[{'LOWER':'puppy'},{'LOWER':'love'}]
pattern3=[{'LOWER':'puppy'},{'IS_PUNCT':True},{'LOWER':'love'}]
matcher.add('PuppyLove',[pattern1,pattern2,pattern3])
```

### Question 2
Find the matches in the doc and print them.

```python
# Find the matches and print them
found_matches=matcher(doc)
print(found_matches)
```

### Question 3
Loop through the matches and print the match id, name, start, end and matched text.

```python
# Print the details of each match
for match_id,start,end in found_matches:
  string_id=nlp.vocab.strings[match_id]
  span=doc[start:end]
  print(match_id,string_id,start,end,span.text)
```

## Setup 2: PhraseMatcher and Text

```python
# Create a PhraseMatcher object
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)
```

```python
# Create doc3 from the text
text='I have a cute kitten and a fluffy puppy. My sleepy cat and happy dog play together.'
doc3 = nlp(text)
```

### Question 4
Make a list of these phrases: `cute kitten`, `fluffy puppy`, `sleepy cat`, `happy dog`.

```python
# Make a list of phrases to match
phrase_list=['cute kitten','fluffy puppy','sleepy cat','happy dog']
```

### Question 5
Convert each phrase in the list into a doc pattern.

```python
# Convert each phrase into a doc pattern
phrase_patterns=[nlp(text) for text in phrase_list]
```

### Question 6
Add the patterns to the phrase matcher. Name the rule `CutePets`.

```python
# Add the patterns to the phrase matcher
matcher.add('CutePets',None,*phrase_patterns)
```

### Question 7
Find the matches in `doc3` and print them.

```python
# Find the matches in doc3 and print them
matches=matcher(doc3)
print(matches)
```
