```python
# Import libraries and load models
import spacy
import nltk
from nltk.stem import WordNetLemmatizer

nlp = spacy.load('en_core_web_sm')

nltk.download('wordnet')
nltk.download('omw-1.4')
lemmatizer = WordNetLemmatizer()
```

### Q1
Lemmatize every token of the sentence and print the original word with its lemma.

```python
# Print each token with its spaCy lemma
text = 'the cats are sleeping in the garden.'
doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

### Q2
Lemmatize these plural nouns.

```python
# Lemmatize plural nouns with spaCy
text = 'cats dogs houses cars'
doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

### Q3
Lemmatize different forms of the verb "play".

```python
# Lemmatize different forms of 'play'
text = 'playing played plays'
doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

### Q4
Lemmatize a multi-sentence text that contains different tenses of the same verb.

```python
# Lemmatize tokens across sentences with different verb tenses
text = 'I am eating.She ate dinner.They have eaten food'
doc = nlp(text)

for token in doc:
  print(token.text, "->", token.lemma_)
```

### Q5
Lemmatize the sentence, remove stop words and punctuation, and keep only the useful lemmas.

```python
# Lemmatize and keep only non-stop, non-punctuation words
text = 'The students are learning data science'
doc = nlp(text)

words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(words)
```

### Q6
Lemmatize a list of plural nouns using the default noun POS.

```python
# Lemmatize plural nouns (default noun POS)
words = ['cats', 'dogs', 'cars', 'houses']

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

### Q7
Lemmatize verb forms by setting `pos='v'`.

```python
# Lemmatize verb forms with pos='v'
words = ['playing', 'played', 'plays']

for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos='v'))
```

### Q8
Lemmatize irregular plural nouns by setting `pos='n'`.

```python
# Lemmatize irregular plural nouns with pos='n'
words = ['children', 'mice', 'geese', 'men', 'women']

for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos='n'))
```

### Q9
Lemmatize adjectives by setting `pos='a'`.

```python
# Lemmatize comparative/superlative adjectives with pos='a'
words = ['better', 'best', 'faster', 'happier']

for word in words:
    print(word, "->", lemmatizer.lemmatize(word, pos='a'))
```

### Q10
Compare how the same word is lemmatized when treated as a noun versus as a verb.

```python
# Compare noun vs verb lemmatization for the same word
word = 'meeting'
print('as noun:', lemmatizer.lemmatize(word, pos='n'))
print('as verb:', lemmatizer.lemmatize(word, pos='v'))
```
